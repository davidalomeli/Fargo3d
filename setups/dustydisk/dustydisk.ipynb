```python
rmin = loadtxt("../../outputs/dustydisk/domain_y.dat")[3:-3]
rmed = 0.5*(rmin[1:]+rmin[:-1])

n = 0

dustvr0 = fromfile("../../outputs/dustydisk/dustvy0.dat")
dustvr = fromfile("../../outputs/dustydisk/dustvy{:d}.dat".format(n))

gasvr0 = fromfile("../../outputs/dustydisk/gasvy0.dat")
gasvr = fromfile("../../outputs/dustydisk/gasvy{:d}.dat".format(n))

cs = fromfile("../../outputs/dustydisk/gasenergy0.dat")
dens = fromfile("../../outputs/dustydisk/gasdens{:d}.dat".format(n))
```

```python
plot(rmed, dustvr0)
plot(rmed, dustvr)

xscale('log')
show()
```

```python
plot(rmed, gasvr0)
plot(rmed, gasvr)

xscale('log')
show()
```

```python
csmed   = (cs[1:]+cs[:-1])*0.5
rhomed  = (dens[1:]+dens[:-1])*0.5
omega   = sqrt(1/rmed**3)
omegamed= (omega[1:]+omega[:-1])*0.5

gradP = (csmed**2)/(rhomed*omegamed) * ( dens[1:]-dens[:-1] )/(rmed[1:]-rmed[:-1])

Ts = 0.02
eps = 0.05

plot( rmed, dustvr-gasvr )
plot(rmed[1:], -Ts**2/(1+Ts**2)*gasvr[1:] + gradP*Ts/((1+eps)**2+Ts**2))
xscale('log')
show()
```
